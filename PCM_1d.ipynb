$$
\begin{align*}
\mathcal{F}[f(x + \Delta x)]
&= \int_{-\infty}^{\infty} f(x + \Delta x) e^{-i\omega x} dx \\
&= \int_{-\infty}^{\infty} f(x') e^{-i\omega(x' - \Delta x)} dx' \\
&= e^{i\omega \Delta x} \mathcal{F}[f(x)] 
\end{align*}
$$

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
base_signal = np.random.uniform(size=1005)
signal_1 = base_signal[:1000]
signal_2 = base_signal[-1000:]

plt.plot(signal_1[:100])
plt.plot(signal_2[:100])
```

```python
signal_1_f = np.fft.fft(signal_1)
signal_2_f = np.fft.fft(signal_2)

plt.plot(np.abs(signal_1_f[:100]))
plt.plot(np.abs(signal_2_f[:100]))

plt.figure()
plt.plot(np.angle(signal_1_f[:100]))
plt.plot(np.angle(signal_2_f[:100]))

plt.figure()
plt.plot(
    np.angle(signal_1_f * np.conjugate(signal_2_f) / np.abs(signal_1_f * signal_2_f))
)
```
